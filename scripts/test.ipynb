```python
import pynapple as nap
import ethograph as eto
from typing import Literal
import matplotlib.pyplot as plt
import numpy as np

path = r"C:\Users\aksel\Documents\AI_data\derivatives\sub-02_id-Poppy\ses-000_date-20260304_01\behav\pynapple\units.npz"
nwb_path = r"C:\Users\aksel\Documents\AI_data\derivatives\sub-02_id-Poppy\ses-000_date-20260304_01\behav\.ethograph\alignment.nwb"


def sliding_window(
    binned: nap.TsdFrame,
    window_size: float,
    step_size: float | None = None,
    reduction: Literal["sum", "mean"] = "mean",
) -> nap.TsdFrame:
    bin_size = binned.time_diff().median(axis=0).round(10)
    window_bins = round(window_size / bin_size)
    kernel = np.ones(window_bins)
    if reduction == "mean":
        kernel /= window_bins
    smoothed = binned.convolve(kernel)
    if step_size is None:
        return smoothed
    return smoothed[:: round(step_size / bin_size)]


tsgroup = nap.load_file(path)
tsgroup

nwb = nap.load_file(nwb_path)
trials_ep = nwb["trials"]

bin_size = 0.005
rate = tsgroup.count(bin_size).restrict(trials_ep) / bin_size
rate_smoothed = sliding_window(rate, window_size=0.025)

# for i in range(5):
#     plt.plot(rate.restrict(trials_ep[i]).d[:, 0])
#     plt.show()

# visualization of features per trial
for i in range(20):
    plt.imshow(rate_smoothed.restrict(trials_ep[i]).d, aspect="auto", cmap="hot", interpolation="nearest")
    plt.colorbar()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np


path = r"C:\Users\aksel\Documents\AI_data\derivatives\sub-02_id-Poppy\ses-000_date-20260304_01\behav\Trial_data3.nc"

import ethograph as eto


dt = eto.open(path)

dt

for trial_num in dt.trials:
    trial = dt.itrial(trial_num)
    dist = trial.pellet_stickClosest_dist
    sim = trial.pellet_stickClosest_angular_similarity

    dist_vals = np.asarray(dist.values).ravel()
    sim_vals = np.asarray(sim.values).ravel()
    time_vals = np.asarray(dist.time).ravel()

    dist_bool = dist_vals < 1
    sim_bool = sim_vals > 0.5
    both_bool = dist_bool & sim_bool

    fig, (ax_dist, ax_sim, ax_both) = plt.subplots(3, 1, sharex=True, figsize=(10, 6))

    dist.plot(ax=ax_dist)
    ax_dist.axhline(1, color="gray", linestyle="--", linewidth=1)
    ax_dist.set_ylabel("distance")

    sim.plot(ax=ax_sim)
    ax_sim.axhline(0.5, color="gray", linestyle="--", linewidth=1)
    ax_sim.set_ylabel("angular sim.")

    for ax in (ax_dist, ax_sim, ax_both):
        ax.fill_between(time_vals, 0, 1, where=both_bool, transform=ax.get_xaxis_transform(),
                        color="green", alpha=0.15, step="mid")

    ax_both.plot(time_vals, both_bool, drawstyle="steps-mid")
    ax_both.set_ylim(-0.1, 1.1)
    ax_both.set_ylabel("both")
    ax_both.set_xlabel("time")

    fig.tight_layout()
    plt.show()
```

```python
from __future__ import annotations

from pathlib import Path

import cv2
import pandas as pd
from movement.io import load_poses


def count_mp4_frames(video_path: Path) -> int:
    capture = cv2.VideoCapture(str(video_path))
    try:
        return int(capture.get(cv2.CAP_PROP_FRAME_COUNT))
    finally:
        capture.release()


def count_pose_frames(pose_path: Path) -> int:
    dataset = load_poses.from_dlc_file(pose_path)
    return int(dataset.sizes["time"])


def find_dlc_file(video_path: Path, dlc_dir: Path) -> Path:
    matches = sorted(dlc_dir.glob(f"{video_path.stem}DLC*.h5"))
    if not matches:
        raise FileNotFoundError(f"No DLC file found for {video_path.name} in {dlc_dir}")
    return matches[0]


def add_frame_counts(
    df: pd.DataFrame,
    video_dir: Path,
    dlc_dir: Path,
    video_columns: tuple[str, ...] = ("video_cam-1", "video_cam-2"),
) -> pd.DataFrame:
    result = df.copy()
    for column in video_columns:
        video_paths = result[column].apply(lambda name: video_dir / name)
        result[f"{column}_mp4_frames"] = video_paths.apply(count_mp4_frames)
        dlc_paths = video_paths.apply(lambda p: find_dlc_file(p, dlc_dir))
        result[f"{column}_dlc_frames"] = dlc_paths.apply(count_pose_frames)
        result[f"{column}_frame_diff"] = (
            result[f"{column}_mp4_frames"] - result[f"{column}_dlc_frames"]
        )
    return result


if __name__ == "__main__":
    video_dir = Path(r"C:\Users\aksel\Documents\VidData\20260421_01_Ivy")
    dlc_dir = Path(r"E:\Backup\Akseli\AI_data\rawdata\sub-01_id-Ivy\ses-000_date-20260421_01\behav\dlc")

    df = nwb.trials.to_dataframe()
    df = add_frame_counts(df, video_dir, dlc_dir)
    print(df)
```
